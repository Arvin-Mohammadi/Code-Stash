# Part I: Reinforcement Learning

## Before discussing reinforcement learning, we will explain the three categories of machine learning problems. These are supervised learning problems, unsupervised learning problems, and reinforcement learning problems.

## [1]. Supervised learning. With supervised learning, we have some input x and some outcome y that we would like to predict with x. So, a supervised learning model is a way of mapping x to y. Thus, our goal is to learn a function that uses x to approximate y. Some supervised learning problems include classification problems like predicting what digit is represented by a handwritten digit made of pixels or predicting whether a given movie review is a positive review or a negative review based on the language in those reviews. So, with classification problems, we are predicting a particular class of outcomes, be it one of the ten classes of handwritten digits or one of the two classes of sentiment positive and negative in that binary classification movie review sentiment problem. The second kind of supervised learning problem is a regression problem. So, with regression problems, instead of predicting a particular category that x belongs to, we are trying to predict some numeric value y given some input x. So, with this, we could predict the future sales of a product or the future value of an asset.


## [2]. Unsupervised learning problems are distinguished by not having any y label available. So, with unsupervised learning problems, we have an input x only. Our goal with an unsupervised learning problem is to learn some hidden underlying structure of the data. 


## [3]. Reinforcement learning problems are ones that we can frame as having an agent that takes some sequence of actions within some environment. The agent could, for example, be a human or an algorithm playing an Atari video game. Alternatively, it could be a human or an algorithm driving a car. Perhaps the primary way in which reinforcement learning problems diverge from supervised or unsupervised ones is that the actions taken by the agent influence the information that the environment provides back to the agent. That is, the agent receives direct feedback on the actions it takes. In contrast, the model never impacts the underlying data with supervised or unsupervised problems. It simply consumes it. For instance, when the agent is playing an Atari video game, the possible actions that can be taken are the buttons that can be pressed on the video game controller. The environment, the Atari console, returns information back to the agent. Another example is when the agent is driving a car. The actions available are much broader and more affluent than for an Atari game in this situation. The agent can adjust the steering column, it can hit the accelerator and the brakes to varying degrees ranging from subtle to dramatic, so those are the kind of actions that the agent can take. Moreover, the environment, in turn, in this case, is the real world, and it consists of roads, traffic, pedestrians, trees, the sky, and so on. This state is then the conditions around the vehicle. So, this is perceived by a human with eyes and ears or by autonomous vehicles cameras and laser radar systems. The reward in the case of an algorithm could be programmed to be positive for every meter of distance traveled towards a destination. You could then take away the reward. In that case, you could have a negative reward for minor traffic infractions and a severe negative reward in the event of a collision.

 # Part VI : Reinforcement Learning Theory
 
 ## [1].The Reinforcement Learning Problems are sequential Decision-Making Problems. They involve an agent taking an action within an environment. Let's say that action is taken at time step t, then the environment returns two types of information back to the agent. It returns a reward. So, this is some scalar numeric value that provides quantitative feedback on the action that the agent took at time t. The agent's objective is to maximize the rewards that it accumulates. And so, rewards are what reinforce productive behaviors that the agent discovers under environmental conditions. State is the second thing that the environment returns in response to any action that the agent takes. So, the state becomes the state at time step t plus one in response to the action that was taken at time step t. Repeating this cycle, the agent, now at time step t plus one, receives a state, and it can, from that state, decide to take some action. And so, this cycle, this Reinforcement Learning Loop repeats over and over. Repeating the above two steps, the agent taking an action, the environment returning the next state, as well as reward-based on that action, this continues to loop until reaching some terminal state. This terminal state could be reached by, for example, obtaining the maximum possible reward. So, in the Cart-Pole Game, that's getting 200 points. Another way that the loop could end would be obtaining some specific desired outcome such as a self-driving car reaching its programmed destination. You could also run out of allotted time, you could use up the maximum number of permitted moves in a game, or the agent could die from encountering a bad guy. The Cart-Pole Game is a popular introductory Reinforcement Learning Problem because it's so simple. So, the Cart-Pole Game has merely four pieces of state information. We have one, the Position of the Cart along the one-dimensional horizontal axis.
 
 ## [2]. The second piece of state information is the Cart's Velocity, so how quickly is it moving along that horizontal axis? Third is Pole Angle, so what is the angle of the pole relative to vertical? Note that if it gets too far away from vertical, that's one of the ways that the game ends. And the fourth and final Environmental State is Pole Angular Velocity, so that's how quickly is the pole moving along its angle. In addition to the state space being quite limited, the action space is limited too. At any given timestep, exactly one action can be taken from only two possible actions. The cart can either move left, or the cart can move right. You can't even stay in place. In this process, this Reinforcement Learning Loop that we've been describing with respect to the Cart-Pole game can be referred to as a Markov Decision Process. What this means is that it's a decision process. So, we have the agent taking actions, making decisions based on state information that it receives from the environment. So, it's a decision process because of that. Markov means that it has the Markov property. This means that we're only using information at the most recent time step to make a decision about what actions we should be taking. So, for example, a trading algorithm that follows the Markov property will only consider the stock prices yesterday, as opposed to trying to consider stock price information from previous days and how that has impacted the most recent days. So, any kind of Markov process is just taking into account the most recent piece of information. Introducing some more terms here, there's a capital letter S, which represents all of the possible states. In The Cart-Pole Game, even with just four kinds of state information, there's still a broad range of possible states. So, the cart could be moving quickly on the right-hand side of the screen with the pole centered and not moving very much, or the cart could be moving slowly on the left side of the screen with the pole moving quite a lot in a particular direction. Even with just four possible pieces of state information, the set of all possible states, capital S, can still be quite large. We use capital A to denote all possible actions. In The Cart-Pole Game, this is relatively simple. It is just the set of movements, right or left. However, with some other problems like a self-driving car, this action space can be quite large. There are different degrees that you could be turning the steering wheel, or different amounts that you could be pressing the gas and the brake.
 
 ## [3].Capital R represents the reward distribution given some pair of some particular state and some particular action. So, we use lowercase s and lowercase a to denote a particular state, and lowercase a to denote a particular action. So, the state that we're in could be having the pole upright and the cart centered, and then the action taken in that circumstance is either a right or a left. The reward that you could expect in a particular situation, it varies and in a lot of environments, there's some unknown hidden reward distribution which our agent needs to learn. So based on a particular state of the environment and some action taken with that state, so the state action, pair, we'll receive some kind of reward. And that distribution initially is mysterious as we take more and more actions in more and more state situations as we accumulate more and more experience, we can get glimpses of what this reward distribution is like. P is the probability of transitioning to some future state given some state, action pair at the current time step. So, if, again, we could have the same state, action pair as we had here, it could be the cart in the center of the screen with the pole positioned vertically above it, and then in that particular circumstance, we take some action, which is moving to the right or moving to the left. Well, this state at time step t plus one. This is a lowercase s here. So, this particular state at time step t plus one, if in The Cart-Pole Game, we take an action of right, then the state at t plus one is going to involve the cart slightly to the right relative to at time step t. So, these transition probabilities prior to any training are unknown to the agent and must be learned. The final term that we need to be aware of in a Markov Decision Process is something called gamma. So, this is the Greek lowercase letter gamma. And this is the discount factor. This discount factor gamma allows us to value immediate rewards more highly than distant rewards. Another function of this discount factor gamma is that it prevents us from looking into the future indefinitely. So, there would be a computational tracked ability problem. If all possible future rewards were valued, then we would be looking ahead infinitely at future time steps. This discount factor gamma means that by 30 time steps into the future, any future reward is worth zero points effectively, and so, this discount factor also gives us a trick to avoid having our computational complexity become too large. The ultimate objective with a Markov Decision Process is to find a function that enables an agent to take an appropriate action given some particular state. In other words, we'd like our agent to learn a function that enables it to map any possible state to any possible action. We call it the Policy function. The high-level idea of the Policy function is this. Regardless of the particular circumstance that the agent finds itself in, what is the policy it should take that will enable it to maximize its future reward. We randomly sample from our memory deque a batch of the size of our mini-batch size that we decided on for stochastic gradient descent. We know precisely what are target reward is. However, if the game is not done, we will calculate the future target reward based on the next state. We will take the reward that we know we have, and then we calculate based on the next state returned. So based on the state at time step T plus one, we use our model to predict the future reward, and we discount that future reward by our discount rate gamma. We will be looking at the results of our agent's performance after every round of gameplay. Note that we are going to be taking exploratory actions all the time. However, as we decay our epsilon rate as we train more and more, we will gradually decrease this epsilon value. Finally, a simple utility method is here for saving model weights at regular intervals. And then, we can choose to load those model weights back in later on. Now We have defined our DQN agent. We are now ready to set up this agent for interacting with an environment.





```python
class DQNAgent:
    def __init__(self, state_size, action_size):
        self.state_size = state_size
        self.action_size = action_size
        self.memory = deque(maxlen=2000) # deque is like list and we can append elements to it.
        self.gamma = 0.95 # discount rate.
        self.epsilon = 1.0 # exploration rate: how much to act randomly.
        self.epsilon_decay = 0.995 # decrease number of random explorations as the agent's performance improves.
        self.epsilon_min = 0.01 # minimum amount of random exploration permitted.
        self.learning_rate = 0.001 # learning rate.
        self.model = self._build_model() # This is a private method 
    
    #==============================================================================================
    #Step 2
    
    def _build_model(self):
        # neural net to approximate Q-value function.# Q is a f of s,a,theta
        model = Sequential()
        model.add(Dense(32, activation='relu', input_dim=self.state_size))
        model.add(Dense(32, activation='relu')) 
        model.add(Dense(self.action_size, activation='linear')) # 2 actions. 2 output neurons: 0 and 1 (L/R)
        model.compile(loss='mse',
                      optimizer=Adam(lr=self.learning_rate)) # You can change this to Adam or other optimisers.
        return model
    
    #==================================================================================================
    
    #Step 3
    
    def remember(self, state, action, reward, next_state, done): # done means game is over or final or early termination.
        self.memory.append((state, action, 
                            reward, next_state, done)) # list of previous experiences, enabling re-training later
     #=====================================================================================================
    
    # Step 4
    
    def train(self, batch_size): # method that trains NN with experiences sampled from memory
        minibatch = random.sample(self.memory, batch_size) # sample a minibatch from memory
        for state, action, reward, next_state, done in minibatch: # extract data for each minibatch sample
            target = reward # if done, then target = reward
            if not done: # if not done, then predict future discounted reward
                target = (reward +  # target here predicts the agent future rewards from the next state
                          self.gamma * # (target) = reward + (discount rate gamma) * # Using Bellman equation
                          np.amax(self.model.predict(next_state)[0])) # (maximum Q value based on future action a')
            target_f = self.model.predict(state) # map current state to our prediction of future discounted reward
            target_f[0][action] = target # We use the discounted future rewards based on the action that the agent took.
            self.model.fit(state, target_f, epochs=1, verbose=0) # using the NN to see which action is optimal,given s 
        if self.epsilon > self.epsilon_min: # epochs=1 b/c no need to use recycle data. We can generate data by OpenAIGym
            self.epsilon *= self.epsilon_decay # Here we are adjusting the value of epsilon if the condition is true.

    def act(self, state):
        if np.random.rand() <= self.epsilon: # if acting randomly, take random action. take a random no from 0 to 1.
            return random.randrange(self.action_size) # The agent at this stage will explore
        act_values = self.model.predict(state) # if not acting randomly, predict reward value based on current state
        # As time progresses, the agent will take less random action=> the agent will exploit.
        # Our model will be used to predict the highest reward based on two actions.
        return np.argmax(act_values[0]) # pick the action that will give maximise the agent reward (left or right)
    
    #============================================================================================================
    
    # Step 5
    # We save our model so we can use it later.
    
    def save(self, name):
        self.model.save_weights(name)

    def load(self, name):
        self.model.load_weights(name)
```

However, we are trying to maximize reward as we are in a reinforcement learning problem in our case. That same algorithm, we call gradient ascent. Because we are not trying to find the lowest cost, we are trying to find the maximum reward. So, the same algorithm, just a little bit of terminology difference. We then save the model weights of our agent so that we can go back and load those in if we want to for some production algorithm. We have built a Deep Reinforcement Learning Agent.

```python
# Step 7


for e in range(n_episodes): # iterate over episodes of gameplay
    
    state = env.reset() # reset state and get four random states.
    state = np.reshape(state, [1, state_size]) # To pass this to the NN.
    
    done = False
    time = 0 # goal is to keep pole upright as long as possible.
    while not done: 
    # This while loop ends if we get to 200 timesteps or the pole falls, or cart disappears from the screen.
        action = agent.act(state) # Given the state, the agent takes an action (move cart left or right).
        next_state, reward, done, _ = env.step(action) # agent interacts with env, gets feedback; 4 state data points.     
        reward = reward if not done else -10 # reward +1 for each additional frame with pole upright and -10 for ending the game.        
        next_state = np.reshape(next_state, [1, state_size]) # To pass this to the NN.
        agent.remember(state, action, reward, next_state, done) # To remember the previous timestep's state, actions, reward, etc.        
        state = next_state # set current state to the current next state for the next while loop       
        if done: # if episode ends: 
            print("episode: {}/{}, score: {}, e: {:.2}" # print the episode's score and agent's epsilon
                  .format(e, n_episodes-1, time, agent.epsilon))
        time += 1
        
        env.render()
        
    if len(agent.memory) > batch_size:
        agent.train(batch_size) # train the agent by replaying the experiences of the episode
    
     # after 50 episodes, we save our weights.
    
    if e % 50 == 0:
        agent.save(output_dir + "weights_" 
                   + '{:04d}'.format(e) + ".hdf5")
        
        
 
```
